# **Medindo a latência de inferência  dos lotes**

```python
def inferencia_ipython(modelo, lotes, desativar_gc=False):
    if desativar_gc:
        gc.disable()

    for i, lote in enumerate(lotes):
        print(f"Medição de tempo para o Lote {i+1}:")
        %timeit -r 7 -n 100 -o modelo.predict(lote)
        print("\n")

    if desativar_gc:
        gc.enable()
```

```python
inferencia_ipython("substitua pelo modelo" , lotes)
```
