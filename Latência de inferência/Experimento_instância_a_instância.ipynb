# **Faz as inferências linha por linha.**

```python
def inferencia_ipython(modelo, previsores_teste, desativar_gc=False):
    if desativar_gc:
        gc.disable()

    for _ in range(50):
        modelo.predict(previsores_teste.iloc[0:1])
    for i in range(len(previsores_teste)):
        amostra = previsores_teste.iloc[i:i+1]
        %timeit -r 7 -n 100 -o modelo.predict(amostra)
    if desativar_gc:
        gc.enable()
```

```python
inferencia_ipython("substitua pelo modelo", previsores_amostra)
```
